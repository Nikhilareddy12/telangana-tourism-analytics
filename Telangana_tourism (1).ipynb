```python
# Calculate total visitors for each year
yearly_visitors = (
    df.groupby("year")["visitors"]
      .sum()
      .reset_index()
)

yearly_visitors
```

Domestic tourist visits in Telangana show considerable variation between
2014 and 2024. Tourism declined sharply during 2020–2021, corresponding to
the COVID-19 period, before recovering strongly from 2022 onward.

The highest recorded annual visitor count in this dataset was approximately
95.2 million in 2016. By 2024, recorded visitors had recovered to approximately
88.2 million, indicating a substantial post-pandemic recovery.

The presence of missing district-month observations means that annual totals
should be interpreted as totals of the available recorded observations.

```python
# Calculate year-over-year percentage growth
yearly_visitors["YoY_Growth_%"] = (
    yearly_visitors["visitors"].pct_change() * 100
)

yearly_visitors
```

```python
# Visualize the annual tourism trend
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=yearly_visitors,
    x="year",
    y="visitors",
    marker="o"
)

plt.title("Annual Domestic Tourist Visits in Telangana (2014–2024)")
plt.xlabel("Year")
plt.ylabel("Number of Visitors")
plt.xticks(yearly_visitors["year"])
plt.grid(True)
plt.show()
```

District-wise Tourism Performance

We aggregate visitor counts by district to identify the districts
that attract the highest number of domestic tourists.

This analysis can help identify major tourism hubs and understand
where tourist activity is concentrated across Telangana.

```python
# Calculate total recorded visitors for each district
district_visitors = (
    df.groupby("district")["visitors"]
      .sum()
      .sort_values(ascending=False)
)

district_visitors.head(10)
```

```python
# Plot the top 10 districts by recorded visitor count
plt.figure(figsize=(10, 6))

district_visitors.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Districts by Domestic Tourist Visits")
plt.xlabel("Number of Visitors")
plt.ylabel("District")
plt.show()
```

```python
#District-wise Tourism Performance in 2024
```

```python
df_2024 = df[df["year"] == 2024]

# Calculate total visitors by district in 2024
district_2024 = (
    df_2024.groupby("district")["visitors"]
           .sum()
           .sort_values(ascending=False)
)

district_2024.head(10)
```

```python
# Plot the top 10 districts in 2024
plt.figure(figsize=(10, 6))

district_2024.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Districts by Domestic Tourist Visits — 2024")
plt.xlabel("Number of Visitors")
plt.ylabel("District")
plt.show()
```

```python
#How much of Telangana's recorded tourism is concentrated in the top 5 districts?
```

```python
total_2024 = district_2024.sum()

top_5_2024 = district_2024.head(5).sum()

top_5_share = (top_5_2024 / total_2024) * 100

top_5_share
```

Tourism activity in 2024 was highly concentrated across a small number of
district/region entries. The five highest-volume entries accounted for
approximately 77% of recorded domestic tourist visits.

The largest entry combines Hyderabad, Ranga Reddy, Medchal-Malkajgiri and
Vikarabad, so it should be interpreted as a combined reporting region rather
than as Hyderabad alone.

This concentration suggests an opportunity for the Telangana Tourism
Department to promote lower-traffic destinations and develop tourism
circuits that distribute visitor activity more broadly across the state.

```python
#COVID-19 Impact and Tourism Recovery
```

```python
# Display annual visitor totals and year-over-year growth
yearly_visitors
```

```python
# Calculate the percentage change between key tourism periods

visitors_2019 = yearly_visitors.loc[
    yearly_visitors["year"] == 2019, "visitors"
].iloc[0]

visitors_2020 = yearly_visitors.loc[
    yearly_visitors["year"] == 2020, "visitors"
].iloc[0]

visitors_2021 = yearly_visitors.loc[
    yearly_visitors["year"] == 2021, "visitors"
].iloc[0]

visitors_2024 = yearly_visitors.loc[
    yearly_visitors["year"] == 2024, "visitors"
].iloc[0]

decline_2019_2020 = (
    (visitors_2020 - visitors_2019) / visitors_2019
) * 100

change_2020_2021 = (
    (visitors_2021 - visitors_2020) / visitors_2020
) * 100

recovery_2021_2024 = (
    (visitors_2024 - visitors_2021) / visitors_2021
) * 100

print(f"2019 → 2020 change: {decline_2019_2020:.2f}%")
print(f"2020 → 2021 change: {change_2020_2021:.2f}%")
print(f"2021 → 2024 change: {recovery_2021_2024:.2f}%")
```

```python
# Compare 2024 tourism with the pre-COVID 2019 level

pre_covid = yearly_visitors.loc[
    yearly_visitors["year"] == 2019, "visitors"
].iloc[0]

latest = yearly_visitors.loc[
    yearly_visitors["year"] == 2024, "visitors"
].iloc[0]

change_2019_2024 = (
    (latest - pre_covid) / pre_covid
) * 100

print(f"2019 → 2024 change: {change_2019_2024:.2f}%")
```

```python
# Visualize the COVID-period decline and subsequent recovery

covid_period = yearly_visitors[
    yearly_visitors["year"].between(2019, 2024)
]

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=covid_period,
    x="year",
    y="visitors",
    marker="o"
)

plt.title("Telangana Domestic Tourism: COVID Impact and Recovery")
plt.xlabel("Year")
plt.ylabel("Recorded Visitors")
plt.xticks(covid_period["year"])
plt.grid(True)
plt.show()
```
